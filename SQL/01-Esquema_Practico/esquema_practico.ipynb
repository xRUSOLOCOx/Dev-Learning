**CONEXION A BASE DE DATOS:**

```python
# Código de conexión a base de datos hello_mysql local:

import mysql.connector
from mysql.connector import Error

conexión = mysql.connector.connect(

host = "localhost",
user = "root",
password = "0000",
database = "hello_mysql"

)

cursor = conexión.cursor()
```

**FUNCIONES DE EJECUCIÓN DE CONSULTAS:**

```python
def select_statements(db_cursor,query):

    db_cursor.execute(query)
    resultado = db_cursor.fetchall()

    return resultado

def modify_statements(db_cursor,query):

    try:

        db_cursor.execute(query)
        conexión.commit()

        return "Correcto"

    except Error as e:
    
        print(e)
```

### **3. EJECUCIÓN DE CONSULTAS BASICAS:**

**SELECT**: Permite obtener datos de una o varias columnas de una tabla.
  - Ejemplo: `SELECT user_name, user_age FROM users;`

**SELECT DISTINCT**: Devuelve solo valores únicos de una columna.
  - Ejemplo: `SELECT DISTINCT user_age FROM users;`

**WHERE**: Filtra los resultados según una condición.
  - Ejemplo: `SELECT * FROM users WHERE user_age > 25;`

**ORDER BY**: Ordena los resultados por una o varias columnas.
  - Ejemplo: `SELECT * FROM users ORDER BY user_surname ASC;`

**LIKE**: Busca patrones en los valores de una columna usando comodines `%`. El símbolo `%` representa cualquier secuencia de caracteres.
  - Ejemplo: `SELECT * FROM hello_mysql.users WHERE user_email LIKE '%@gmail.com';`

**IS NULL**: Selecciona filas donde el valor de la columna es nulo.
  - Ejemplo: `SELECT * FROM users WHERE user_surname IS NULL;`

**IS NOT NULL**: Selecciona filas donde el valor de la columna no es nulo.
  - Ejemplo: `SELECT * FROM users WHERE user_surname IS NOT NULL;`

**MIN()**: Devuelve el valor mínimo de una columna numérica.
  - Ejemplo: `SELECT MIN(user_age) FROM hello_mysql.users;`

**MAX()**: Devuelve el valor máximo de una columna numérica.
  - Ejemplo: `SELECT MAX(user_age) FROM hello_mysql.users;`

**SUM()**: Calcula la suma de todos los valores de una columna numérica.
  - Ejemplo: `SELECT SUM(user_age) FROM hello_mysql.users;`

**AVG()**: Calcula el promedio (media aritmética) de una columna numérica.
  - Ejemplo: `SELECT AVG(user_age) FROM hello_mysql.users;`

**COUNT()**: Cuenta el número de filas que cumplen una condición o el total si se usa `COUNT(*)`.
  - Ejemplo: `SELECT COUNT(*) FROM hello_mysql.users;`
  - Ejemplo condicional: `SELECT COUNT(user_id) FROM hello_mysql.users WHERE user_age > 30;`

**GROUP BY**: Se usa para agrupar filas que tienen los mismos valores en una o más columnas.

  - Ejemplo: `SELECT user_surname, COUNT(user_age) FROM hello_mysql.users GROUP BY user_surname;`

**CASE END**: Se usa para evaluar multiples condiciones y obtener un resultado de estas.

  - Ejemplo: `SELECT *, CASE WHEN user_age > 17 THEN 'Eres mayor de edad' ELSE 'Eres menor de edad' END AS 'Test de edad' FROM users`

**IFNULL()**. Se usa en caso de querer reemplazar los valores nulos en una columna por un valor deseado.

  - Ejemplo: `SELECT user_name,user_surname,IFNULL(user_age,0) FROM hello_mysql.users`
